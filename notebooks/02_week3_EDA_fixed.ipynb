```python
# Week 3 - Exploratory Data Analysis (Tanmeet)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/cleaned/cleaned_data.csv')

# Preview data
df.head()
```

```python
# Check missing values
print("Missing values per column:")
print(df.isnull().sum())

# Check duplicates
print("\nDuplicate rows:", df.duplicated().sum())

# Data types summary
print("\nData types:")
print(df.dtypes)
```

```python
df = df.drop_duplicates()
```

```python
# Summary of numerical features
df.describe()

# Unique value counts for categorical columns
print("\nUnique Careers:")
print(df['Career'].value_counts())

```

### 📘 Data Dictionary

| Column Name | Description | Data Type |
|--------------|-------------|-----------|
| O_score | Openness personality score | float |
| C_score | Conscientiousness personality score | float |
| E_score | Extraversion personality score | float |
| A_score | Agreeableness personality score | float |
| N_score | Neuroticism personality score | float |
| Numerical Aptitude | Numerical reasoning ability | float |
| Spatial Aptitude | Spatial reasoning ability | float |
| Perceptual Aptitude | Attention to visual patterns | float |
| Abstract Reasoning | Logical thinking and pattern solving | float |
| Verbal Reasoning | Vocabulary and language reasoning | float |
| Career | Target job role label | object |


```python
# Summary statistics
print("📊 Summary of numerical columns:")
display(df.describe())

# Check overall averages per career
career_means = df.groupby('Career').mean(numeric_only=True)
print("\n🎯 Average scores for each career:")
display(career_means)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Histogram of one aptitude score
plt.figure(figsize=(7,4))
sns.histplot(df['Numerical Aptitude'], kde=True, color='skyblue')
plt.title('Distribution of Numerical Aptitude Scores')
plt.show()

# Career distribution
plt.figure(figsize=(8,4))
df['Career'].value_counts().plot(kind='bar', color='orange')
plt.title('Career Count Distribution')
plt.ylabel('Number of Students')
plt.xlabel('Career')
plt.show()
```
